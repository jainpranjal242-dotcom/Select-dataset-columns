## Objective
Learn to access specific rows and columns using Pandas `loc` and `iloc`.

```python
import pandas as pd

df = pd.read_csv('iris_dataset.csv')
df.head()
```

## Examples using `loc`

```python
df.loc[:, 'species'].head()
```

```python
df.loc[:, ['sepal_length_cm', 'petal_length_cm', 'species']].head()
```

```python
df.loc[0:4, ['sepal_length_cm', 'sepal_width_cm', 'species']]
```

```python
df.loc[df['petal_length_cm'] > 5, ['petal_length_cm', 'petal_width_cm', 'species']]
```

## Examples using `iloc`

```python
df.iloc[:, 0].head()
```

```python
df.iloc[:, [0, 2, 4]].head()
```

```python
df.iloc[0:5, 0:3]
```

```python
df.iloc[10:15, 0:5]
```

## Key Observations

1. `loc` primarily uses labels.
2. `iloc` primarily uses integer positions.
3. Both methods can select rows and columns together.
4. `loc` can be combined with Boolean filtering.
5. Selecting required columns improves readability.
6. `iloc` uses zero-based positions.
7. Meaningful column labels make `loc` especially readable.

## Conclusion
Pandas `loc` and `iloc` are essential tools for selecting and filtering DataFrame data.
